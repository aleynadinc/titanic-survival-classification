# Titanic - Hayatta Kalma Tahmini (Lojistik Regresyon)

## Verileri içe aktarın

```python
import pandas as pd

df = pd.read_csv('https://drive.google.com/uc?export=download&id=1mZEvxp_MRAeE89h2CU4XD02n6XXTt-5f')
df.head()
```

## Verileri inceleyin

```python
df.info()
```

```python
df.describe()
```

Hedef: `Survived` (0 = öldü, 1 = hayatta kaldı). Potansiyel özellikler: `Pclass`, `Sex`, `Age`, `SibSp`, `Parch`, `Fare`, `Embarked`.

## X ve y'yi seçin

```python
X = df[['Pclass', 'SibSp', 'Parch', 'Fare']]
y = df['Survived']
```

## Train Test Bölümü

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=0)
```

## Normalleştirme

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

## Modelleme

```python
from sklearn.linear_model import LogisticRegression

model = LogisticRegression()
model.fit(X_train_scaled, y_train)
```

## Model Değerlendirmesi

```python
model.score(X_test_scaled, y_test)
```

## Tahmin

```python
new_passenger = pd.DataFrame([{
    'Pclass': 3,
    'SibSp': 0,
    'Parch': 2,
    'Fare': 15.5
}])

new_passenger_scaled = scaler.transform(new_passenger)
model.predict(new_passenger_scaled)
```

**Sonucu nasıl yorumluyorsunuz?**

Tahmin sonucu 0 ise yolcunun hayatını kaybettiğini, 1 ise hayatta kaldığını gösteriyor. Bu yolcunun düşük sınıf bileti ve düşük ücretiyle hayatta kalma şansı düşük görünüyor.
